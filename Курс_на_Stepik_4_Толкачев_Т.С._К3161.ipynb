**Задание 2. Функция Гаусса**

Напишите свою реализацию гауссовского фильтра. Сделайте это в несколько шагов. В этом задании выполняется первый шаг.

Напишите функцию для создания ядра гауссовского фильтра. На вход функция принимает $ \sigma $ и возвращает квадратную матрицу.

Функция Гаусса:
> $G(r)=\frac{1}{2\pi\sigma^2}e^{-\frac{r^2}{2\sigma^2}}$,
где $r=\sqrt{(x-x_0)^2 + (y-y_0)^2}$, $x_0, y_0$ -- центр фильтра


Для рассчета размера ядра воспользуйтесь формулой: $k = 3*\sigma$

```python
def gauss(r, sigma):
  return (1/(2*np.pi*sigma**2)) * (np.exp(-(r**2)/(2*sigma**2)))


def gauss_kernel(sigma):
  k = int(3 * sigma)
  kernel = np.zeros((k, k))

  for y in range(k):
    for x in range(k):
      kernel[y][x] = gauss(np.sqrt((x - k//2)**2 + (y - k//2)**2), sigma)

  return kernel / kernel.sum()
```


**Задание 3. Операция свертки**

Напишите свою реализацию функцию для реализации операции свертки. В качестве теста сгенерируйте ядро box-фильтра и примените к картинке из задания 1.

```python
def add_padding(image, pix=1):
  return np.pad(
      image,
      pad_width=((pix, pix), (pix, pix), (0, 0)),
      mode='constant'
  )

def convolve(image, kernel, stride = 1):
  assert kernel.shape[0] == kernel.shape[1] and kernel.shape[0] % 2, "Invalid kernel dimensions"

  k = kernel.shape[0]//2
  h, w = image.shape[:2]

  conv = image.copy()
  image = add_padding(image, k)

  n_channels = image.shape[2] if len(image.shape) > 2 else 1

  for i in range(k, h-k, stride):
    for j in range(k, w-k, stride):
      for c in range(n_channels):
        conv[i, j, c] = np.clip(
          (kernel * image[i-k:i+k+1, j-k:j+k+1, c]).sum(), 0, 255)

  return conv
```

```python
kernel_size = 5
my_kernel = np.ones((kernel_size, kernel_size)) / (kernel_size ** 2)

box_img = convolve(img, my_kernel)
showmany(img, box_img)
```

**Задание 4. Фильтр Гаусса**

Соедините задания 2 и 3. То есть напишите код, который сначала генерирует ядро гауссовского фильтра, а затем применяет этот фильтр к изображению. Протестируйте на нескольких изображениях

```python
def guass_filter(image, sigma=1):
  kernel = gauss_kernel(sigma)
  return convolve(image, kernel)
```

```python
def guass_sharpen(image, radius=1, amount=1.5):
  blurred = guass_filter(image, radius)
  return np.clip(image + amount * (image - blurred), 0, 255).astype(np.uint8)
```

```python
def get_hist(image):
  return np.histogram(image, bins=range(257))[0]

def sub_hist(hist, image):
  for pix in image:
    hist[pix] = np.clip(hist[pix]-1, 0, 255)

def add_hist(hist, image):
  for pix in image:
    hist[pix] = np.clip(hist[pix]+1, 0, 255)

def median(histogram, k):
  counter = 0
  for i, v in enumerate(histogram):
    counter += v
    if counter > k**2/2:
      return i

def median_filter(image, k = 7):
  h, w = image.shape[:2]
  result = np.zeros((image.shape[0]-k+1, image.shape[1]-k+1))

  for i in range(h-k):
    hist = get_hist(image[h:h+k, :k])
    for j in range(w-k):
      result[i, j] = median(hist, k)
      sub_hist(hist, image[i:i+k, j])
      add_hist(hist, image[i:i+k, j+k])

  return np.clip(result, 0, 255).astype(np.uint8)
```

```python
img = imread('img.png')
new_img = median_filter(img)
```
