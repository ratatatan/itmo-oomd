**Задание 2. Функция Гаусса**

Напишите свою реализацию гауссовского фильтра. Сделайте это в несколько шагов. В этом задании выполняется первый шаг.

Напишите функцию для создания ядра гауссовского фильтра. На вход функция принимает $ \sigma $ и возвращает квадратную матрицу.

Функция Гаусса:
> $G(r)=\frac{1}{2\pi\sigma^2}e^{-\frac{r^2}{2\sigma^2}}$,
где $r=\sqrt{(x-x_0)^2 + (y-y_0)^2}$, $x_0, y_0$ -- центр фильтра


Для рассчета размера ядра воспользуйтесь формулой: $k = 3*\sigma$

```python
def gauss(r, sigma):
  return (1/(2*np.pi*sigma**2)) * (np.exp(-(r**2)/(2*sigma**2)))


def gauss_kernel(sigma):
  k = int(3 * sigma)
  kernel = np.zeros((k, k))

  for y in range(k):
    for x in range(k):
      kernel[y][x] = gauss(np.sqrt((x - k//2)**2 + (y - k//2)**2), sigma)

  return kernel / kernel.sum()
```


**Задание 3. Операция свертки**

Напишите свою реализацию функцию для реализации операции свертки. В качестве теста сгенерируйте ядро box-фильтра и примените к картинке из задания 1.

```python
def add_padding(image, pix=1):
  h, w = image.shape[0], image.shape[1]
  n_channels = image.shape[2] if len(image.shape) > 2 else 1

  canvas = np.dstack([np.zeros((h+pix*2, w+pix*2))] * n_channels)
  canvas[pix:-pix, pix:-pix] = image

  return canvas

def convolve(image, kernel, stride = 1):
  conv = image.copy()
  image = add_padding(image)
  ih, iw = image.shape[0], image.shape[1]
  kh, kw = kernel.shape[0]//2, kernel.shape[1]//2
  odd = kernel.shape[0] % 2
  n_channels = image.shape[2] if len(image.shape) > 2 else 1

  for i in range(kh, ih-kh-odd, stride):
    for j in range(kw, iw-kw-odd, stride):
      for c in range(n_channels):
        conv[i, j, c] = np.clip(
          (kernel * image[i-kh:i+kh+odd, j-kw:j+kw+odd, c]).sum(), 0, 255)

  return conv

kernel_size = 5
my_kernel = np.ones((kernel_size, kernel_size)) / (kernel_size ** 2)

box_img = convolve(img, my_kernel)
showmany(img, box_img)
```

**Задание 4. Фильтр Гаусса**

Соедините задания 2 и 3. То есть напишите код, который сначала генерирует ядро гауссовского фильтра, а затем применяет этот фильтр к изображению. Протестируйте на нескольких изображениях

```python
def gauss_filter(image, sigma=1.0):
  kernel = gauss_kernel(sigma)
  return convolve(image, kernel)

showmany(img, gauss_filter(img, 5.0))
```
